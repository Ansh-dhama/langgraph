```python
from langgraph.graph import StateGraph , START , END
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
```

```python
load_dotenv()

model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)
```

```python
class BLOG(TypedDict):
    title: str
    outline: str
    content: str
```

```python
def create_outline(state: BLOG) -> BLOG:

    title = state['title']

    prompt = f'Generate an outline for a blog on the topic - {title}'

    outline = model.invoke(prompt).text

    state['outline'] = outline

    return state
```

```python
def create_blog(state: BLOG)-> BLOG:

    topic= state['title']

    outline = state['outline']

    propmt =f'write a detaile blog on the topic {topic} using the following outline {outline}'

    content = model.invoke(propmt).text

    state['content']=content

    return state
```

```python
graph = StateGraph(BLOG)

graph.add_node('create_blog',create_blog)
graph.add_node('create_outline' , create_outline)

graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline', 'create_blog')
graph.add_edge('create_blog' , END)
graph.compile
workflow = graph.compile()
```

```python
intial = {'title' : 'Rise of AI in India'}

final = workflow.invoke(intial)

print(final)
```
